```python
from collections import deque

graph = {
    'A': ['B' , 'C'],
    'B': ['D' , 'E'],
    'C': ['F' , 'G'],
    'D': ['H'],
    'E': [],
    'F': ['I'],
    'G': ['J'],
    'H': [],
    'I': [],
    'J': []
}
def bsf(graph, start):
  visited = []
  queue = deque()

  visited.append(start)
  queue.append(start)

  while queue:

    node = queue.popleft();
    print(node , end=' ')

    for neighbour in graph[node]:
      if neighbour not in visited:
        visited.append(neighbour)
        queue.append(neighbour)

start = input ("Enter start node: ").upper()

print("BSF Traversal: ")
bsf(graph, start)
```

```python
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': ['H'],
    'E': [],
    'F': ['I', 'J'],
    'G': [],
    'H': [],
    'I': [],
    'J': []
}

def dfs(graph, start):
    visited = []
    stack = [start]

    while stack:
        node = stack.pop()

        if node not in visited:
            print(node, end=" ")
            visited.append(node)

            for neighbour in reversed(graph[node]):
                stack.append(neighbour)

start = input("Enter start node: ").upper()

print("DFS Traversal:")
dfs(graph, start)
```
