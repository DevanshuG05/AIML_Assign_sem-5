```python
import heapq

# Graph is already defined in the program
graph = {
    'A': {'B': 1, 'C': 3},
    'B': {'D': 3, 'E': 1},
    'C': {'F': 2},
    'D': {'G': 2},
    'E': {'G': 2},
    'F': {'G': 1},
    'G': {}
}

# Heuristic values
# Estimated cost from each node to goal G
heuristic = {
    'A': 6,
    'B': 4,
    'C': 4,
    'D': 2,
    'E': 2,
    'F': 1,
    'G': 0
}


def a_star(start, goal):

    # Priority queue: (f_cost, node)
    open_list = []
    heapq.heappush(open_list, (0, start))

    # Cost from start node
    g_cost = {start: 0}

    # To store the path
    parent = {start: None}

    while open_list:

        # Get node having minimum f(n)
        current_f, current = heapq.heappop(open_list)

        # If goal is reached
        if current == goal:

            path = []

            while current is not None:
                path.append(current)
                current = parent[current]

            path.reverse()

            return path, g_cost[goal]

        # Explore neighboring nodes
        for neighbor, edge_cost in graph[current].items():

            # Calculate g(n)
            new_g = g_cost[current] + edge_cost

            # If a better path is found
            if neighbor not in g_cost or new_g < g_cost[neighbor]:

                g_cost[neighbor] = new_g

                # f(n) = g(n) + h(n)
                f = new_g + heuristic[neighbor]

                heapq.heappush(open_list, (f, neighbor))

                parent[neighbor] = current

    return None, None


# -----------------------------
# USER INPUT
# -----------------------------

print("Available Nodes:", list(graph.keys()))

start = input("Enter Start Node: ").upper()
goal = input("Enter Goal Node: ").upper()


# Validate input
if start not in graph or goal not in graph:
    print("Invalid node entered.")

else:

    path, cost = a_star(start, goal)

    if path:
        print("\nA* Algorithm Result")
        print("-------------------")
        print("Start Node :", start)
        print("Goal Node  :", goal)
        print("Shortest Path :", " -> ".join(path))
        print("Total Cost :", cost)

    else:
        print("No path exists between", start, "and", goal)
```
