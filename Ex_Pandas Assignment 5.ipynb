```python
import math
import numpy as np
from scipy import stats, optimize, integrate

print("=== 1. MATH MODULE ===")
num = 16.75
print(f"Floor: {math.floor(num)}, Ceil: {math.ceil(num)}")
print(f"Factorial of 5: {math.factorial(5)}")
print(f"GCD of (24, 60): {math.gcd(24, 60)}")
print(f"Hypotenuse (3, 4): {math.hypot(3, 4)}")

print("\n=== 2. NUMPY ===")
a = np.array([10, 20, 30, 40, 50])
print(f"Array: {a}")
print(f"Mean: {a.mean()}, Standard Deviation: {a.std():.2f}")
print(f"Vectorized Operation (a * 2 + 5): {a * 2 + 5}")


A = np.array([[2, 1], 
              [1, 3]])
b = np.array([8, 14])
solution = np.linalg.solve(A, b)
print(f"Linear system solution: x = {solution[0]}, y = {solution[1]}")

print("\n=== 3. SCIPY ===")

integral_val, _ = integrate.quad(lambda x: x**2, 0, 3)
print(f"Integral of x^2 over [0, 3]: {integral_val}")

min_res = optimize.minimize(lambda x: (x - 3)**2 + 5, x0=[0])
print(f"Minimum found at x = {min_res.x[0]:.2f} with value {min_res.fun:.2f}")

# Statistics: Normal Distribution CDF at Z = 1.96
print(f"Standard Normal CDF at z=1.96: {stats.norm.cdf(1.96):.4f}")
```
