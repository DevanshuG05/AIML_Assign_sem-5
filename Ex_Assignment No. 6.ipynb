```python
import pandas as pd

file_path = r"D:\AIML LAB\iris\iris.data"


columns = [
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width",
    "species"
]

df = pd.read_csv(file_path, header=None, names=columns)

print("✓ Dataset loaded successfully!")
print(f"Rows    : {df.shape[0]}")
print(f"Columns : {df.shape[1]}")
```

```python
print("Statistical Summary")
print("=" * 80)

display(df.describe())
```

```python
print("Missing Values")
print("=" * 80)

missing_values = df.isnull().sum()

display(missing_values.to_frame(name="Missing Count"))
```

```python

df_preprocessed = df.copy()

df_preprocessed.loc[0, "sepal_length"] = np.nan
df_preprocessed.loc[5, "sepal_width"] = np.nan
df_preprocessed.loc[10, "petal_width"] = np.nan

print("Missing values introduced for demonstration:")
print("=" * 60)

display(df_preprocessed.isnull().sum().to_frame(name="Missing Count"))
```

```python

numeric_columns = [
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width"
]

for column in numeric_columns:
    df_preprocessed[column] = df_preprocessed[column].fillna(
        df_preprocessed[column].mean()
    )

print("Missing values after mean imputation:")
print("=" * 60)

display(df_preprocessed.isnull().sum().to_frame(name="Missing Count"))
```

```python
print("Dataset after handling missing values")
print("=" * 80)

display(df_preprocessed.head(10))
```

```python

df_preprocessed = df.copy()

print("✓ Preprocessing copy created successfully.")
print(f"Original dataset shape    : {df.shape}")
print(f"Preprocessing dataset shape: {df_preprocessed.shape}")
```

```python

df_preprocessed.loc[0, "sepal_length"] = np.nan
df_preprocessed.loc[5, "sepal_width"] = np.nan
df_preprocessed.loc[10, "petal_width"] = np.nan

print("Missing values introduced for demonstration")
print("=" * 60)

display(df_preprocessed.isnull().sum().to_frame(name="Missing Count"))
```

```python

numeric_columns = [
    "sepal_length",
    "sepal_width",
    "petal_length",
    "petal_width"
]

for column in numeric_columns:
    df_preprocessed[column] = df_preprocessed[column].fillna(
        df_preprocessed[column].mean()
    )

print("✓ Missing values handled using Mean Imputation")
print("=" * 60)

display(df_preprocessed.isnull().sum().to_frame(name="Missing Count"))
```

```python
print("Categorical Column")
print("=" * 60)

print("Column:", "species")
print("\nUnique categories:")

for category in df_preprocessed["species"].unique():
    print("•", category)
```

```python
from sklearn.preprocessing import MinMaxScaler


scaler = MinMaxScaler()

df_preprocessed[numeric_columns] = scaler.fit_transform(
    df_preprocessed[numeric_columns]
)

print("✓ Numerical features normalized using Min-Max Scaling")
print("=" * 60)
```

```python
print("Dataset After Normalization")
print("=" * 80)

display(df_preprocessed.head(10))
```

```python
print("Minimum Values After Normalization")
print("=" * 60)

display(df_preprocessed[numeric_columns].min().to_frame(name="Minimum"))

print("\nMaximum Values After Normalization")
print("=" * 60)

display(df_preprocessed[numeric_columns].max().to_frame(name="Maximum"))
```

```python
comparison = pd.DataFrame({
    "Feature": numeric_columns,
    "Original Mean": [
        df[column].mean() for column in numeric_columns
    ],
    "Normalized Mean": [
        df_preprocessed[column].mean() for column in numeric_columns
    ]
})

display(comparison)
```
