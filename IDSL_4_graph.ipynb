```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# For reproducible results
np.random.seed(42)
```

```python
print("========== 1. EVENT DEPENDENCE AND INDEPENDENCE ==========")

# Simulate two dice rolls
A = np.random.randint(1, 7, 1000)
B = np.random.randint(1, 7, 1000)
```

```python
P_A = np.mean(A > 3)
P_B = np.mean(B % 2 == 0)

# Probability of A and B occurring together
P_A_and_B = np.mean((A > 3) & (B % 2 == 0))
```

```python
# Conditional probability P(A|B)
P_A_given_B = P_A_and_B / P_B

print("P(A):", P_A)
print("P(B):", P_B)
print("P(A ∩ B):", P_A_and_B)
print("P(A | B):", P_A_given_B)
```

```python
# Check independence
independent = np.isclose(P_A_and_B, P_A * P_B, atol=0.05)

print("P(A) × P(B):", P_A * P_B)
print("Are A and B independent?", independent)
```

```python
# Create contingency table
data = {
    'Passed Math': [30, 20],
    'Failed Math': [10, 40]
}
table = pd.DataFrame(
    data,
    index=['Passed English', 'Failed English']
)
print("\nContingency Table:")
print(table)
```

```python
# P(Passed Math | Passed English)
P_PM_given_PE = (
    table.loc['Passed English', 'Passed Math']
    / table.loc['Passed English'].sum()
)

print("\nP(Passed Math | Passed English):", P_PM_given_PE)
```

```python
# Display as bar graph
table.plot(kind='bar')
plt.title("Math and English Results")
plt.xlabel("English Result")
plt.ylabel("Number of Students")
plt.xticks(rotation=0)
plt.show()
```

```python
print("\n========== 3. BAYES' THEOREM ==========")

# Given probabilities
P_spam = 0.01
P_not_spam = 0.99

P_positive_given_spam = 0.99
P_positive_given_not_spam = 0.05
```

```python
# Total probability of a positive result
P_positive = (
    P_positive_given_spam * P_spam
    + P_positive_given_not_spam * P_not_spam
)
```

```python
# Bayes' theorem
P_spam_given_positive = (
    P_positive_given_spam * P_spam
) / P_positive

print("P(Spam):", P_spam)
print("P(Not Spam):", P_not_spam)
print("P(Positive | Spam):", P_positive_given_spam)
print("P(Positive | Not Spam):", P_positive_given_not_spam)
print("P(Positive):", P_positive)
print("P(Spam | Positive):", P_spam_given_positive)

print(
    "\nPercentage of emails that are actually spam "
    "when the test is positive:",
    P_spam_given_positive * 100, "%"
)
```

```python
print("\n========== 4. RANDOM VARIABLES & CONTINUOUS DISTRIBUTION ==========")

# Generate 1000 random values from a normal distribution
data = np.random.normal(
    loc=50,
    scale=10,
    size=1000
)
```

```python
# Calculate mean
mean = np.mean(data)
```

```python
# Calculate standard deviation
std_dev = np.std(data)
```

```python
# Probability P(40 < X < 60)
prob = np.mean((data > 40) & (data < 60))
```

```python
print("Mean:", mean)
print("Standard Deviation:", std_dev)
print("P(40 < X < 60):", prob)
print("Percentage:", prob * 100, "%")
```

```python
# Plot histogram
plt.figure(figsize=(8, 5))
plt.hist(data, bins=30, edgecolor='black')
plt.title("Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.axvline(40, linestyle='--', label='X = 40')
plt.axvline(60, linestyle='--', label='X = 60')
plt.legend()
plt.show()
```

```python
# Create a non-normal population
population = np.random.exponential(
    scale=2,
    size=10000
)
```

```python
# Generate sample means
sample_means = [
    np.mean(np.random.choice(population, 30))
    for _ in range(1000)
]
```

```python
# Calculate statistics
mean_sample_means = np.mean(sample_means)
std_sample_means = np.std(sample_means)

print("Population Mean:", np.mean(population))
print("Mean of Sample Means:", mean_sample_means)
print("Standard Deviation of Sample Means:", std_sample_means)
```

```python
# Plot original exponential population
plt.figure(figsize=(8, 5))
plt.hist(population, bins=40, edgecolor='black')
plt.title("Original Population - Exponential Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

```python
# Plot distribution of sample means
plt.figure(figsize=(8, 5))
plt.hist(sample_means, bins=30, edgecolor='black')
plt.title("Central Limit Theorem - Distribution of Sample Means")
plt.xlabel("Sample Mean")
plt.ylabel("Frequency")
plt.show()
```
