```python
from ucimlrepo import fetch_ucirepo
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
```

```python
adult = fetch_ucirepo(id=2)
X = adult.data.features
y = adult.data.targets

print("Dataset loaded successfully!")
print("Original shape:", X.shape)
```

```python
print("\nFirst 5 rows:")
display(X.head())
```

```python
X = X.replace("?", np.nan)
```

```python
print("\nMissing values before handling:")
print(X.isnull().sum())
```

```python
numerical_columns = X.select_dtypes(include=['int64', 'float64']).columns
categorical_columns = X.select_dtypes(include=['object', 'str']).columns

# Fill numerical missing values with median
for column in numerical_columns:
    X[column] = X[column].fillna(X[column].median())

# Fill categorical missing values with mode
for column in categorical_columns:
    X[column] = X[column].fillna(X[column].mode()[0])

print("Missing values after handling:")
print(X.isnull().sum())
```

```python
from sklearn.preprocessing import OneHotEncoder

# One-hot encode categorical columns
X_encoded = pd.get_dummies(
    X,
    columns=categorical_columns,
    drop_first=True
)

print("Shape after encoding:")
print(X_encoded.shape)

display(X_encoded.head())
```

```python
from sklearn.preprocessing import MinMaxScaler

# Identify numerical columns after encoding
numerical_columns = X.select_dtypes(
    include=['int64', 'float64']
).columns

# Create scaler
scaler = MinMaxScaler()

# Normalize numerical columns
X_encoded[numerical_columns] = scaler.fit_transform(
    X_encoded[numerical_columns]
)

print("Normalized data:")
display(X_encoded.head())
```

```python
print("Final dataset shape:")
print(X_encoded.shape)

print("\nMissing values:")
print(X_encoded.isnull().sum().sum())

print("\nMinimum values:")
print(X_encoded[numerical_columns].min())

print("\nMaximum values:")
print(X_encoded[numerical_columns].max())
```
